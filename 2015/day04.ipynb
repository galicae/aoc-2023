```python
import hashlib
from tqdm import tqdm
```

```python
def naive_finder(secret, rfrom=0, rto=1000000000, leading_zeroes=5):
    for number in tqdm(range(rfrom, rto)):
        s = secret + str(number)
        res = hashlib.md5(s.encode())
        md5_hash = res.hexdigest()
        if md5_hash[:leading_zeroes] == "0" * leading_zeroes:
            return number
    print("none found, increase range")
```

```python
secret = "abcdef"
assert naive_finder(secret) == 609043
```

```python
secret = "pqrstuv"
assert naive_finder(secret) == 1048970
```

```python
secret = "yzbqklnj"
naive_finder(secret)
```
