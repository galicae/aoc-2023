```python
import numpy as np
from util import file_to_list
```

```python
def parse_instruction(instruction):
    words = instruction.split()
    match words[0]:
        case "turn":
            do = words[1]
            x1, y1 = [int(i) for i in words[2].split(",")]
            x2, y2 = [int(i) for i in words[4].split(",")]
        case "toggle":
            do = words[0]
            x1, y1 = [int(i) for i in words[1].split(",")]
            x2, y2 = [int(i) for i in words[3].split(",")]

    return do, x1, y1, x2, y2
```

```python
def santa_instruction(grid, instruction, n=1000):
    do, x1, y1, x2, y2 = parse_instruction(instruction)

    # turn on:
    match do:
        case "on":
            grid[x1:x2+1, y1:y2+1] = True
        case "toggle":
            grid[x1:x2+1, y1:y2+1] = ~grid[x1:x2+1, y1:y2+1]
        case "off":
            grid[x1:x2+1, y1:y2+1] = False
```

```python
santa = file_to_list("./data/day06.txt")
n = 1000
grid = np.zeros((n, n), dtype=bool)

for s in santa:
    santa_instruction(grid, s, n=10)

print(grid.sum())
```

## quantitative lights

```python
def santa_instruction(grid, instruction, n=1000):
    do, x1, y1, x2, y2 = parse_instruction(instruction)

    # turn on:
    match do:
        case "on":
            grid[x1:x2+1, y1:y2+1] += 1
        case "toggle":
            grid[x1:x2+1, y1:y2+1] += 2
        case "off":
            grid[x1:x2+1, y1:y2+1] -= 1
            grid[grid<0] = 0
```

```python
n = 1000
grid = np.zeros((n, n))

for s in santa:
    santa_instruction(grid, s, n=10)

print(grid.sum())
```
