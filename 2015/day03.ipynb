# infinite grid houses

For example:
```
- > delivers presents to 2 houses: one at the starting location, and one to the east.
- ^>v< delivers presents to 4 houses in a square, including twice to the house at his starting/ending location.
- ^v^v^v^v^v delivers a bunch of presents to some very lucky children at only 2 houses.
```

````python
def follow(x, y, c):
    match c:
        case ">":
            return x + 1, y
        case "<":
            return x - 1, y
        case "^":
            return x, y + 1
        case "v":
            return x, y - 1

def initialize_grid():
    houses = set()
    start = (0, 0)
    houses.add(start)
    return houses

def google_maps(directions):
    houses = initialize_grid()
    x, y = [0, 0]
    for d in directions:
        house = follow(x, y, d)
        houses.add(tuple(house))
        x, y = house
    return houses
````

````python
# initialize visited houses
directions = ">"
visited = google_maps(directions)
print(len(visited))

# initialize visited houses
directions = "^>v<"
visited = google_maps(directions)
print(len(visited))

# initialize visited houses
directions = "^v^v^v^v^v"
visited = google_maps(directions)
print(len(visited))
````
