# part 1

```python
data = file_to_list("./data/day02.txt")
```

```python
def parse_size(line):
    dim = [int(num) for num in line.split("x")]
    return dim

def area(dim):
    l, w, h = dim
    # first the main area
    main_area = 2*l*w + 2*w*h + 2*h*l
    # now remove the biggest coordinates
    dim_sorted = np.sort(dim)
    slack = dim_sorted[0] * dim_sorted[1]

    return main_area + slack
```

```python
total_area = 0
for line in data:
    dim = parse_size(line)
    total_area += area(dim)

print(total_area)
```

```python
# ribbon length for a single present:
def ribbon(dim):
    # area of smallest side
    dim_sorted = np.sort(dim)
    slack = dim_sorted[0] * 2 + dim_sorted[1] * 2
    # volume of present:
    volume = dim[0] * dim[1] * dim[2]

    return slack + volume
```

```python
total_ribbon = 0
for line in data:
    dim = parse_size(line)
    total_ribbon += ribbon(dim)

print(total_ribbon)
```
