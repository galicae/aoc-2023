# Little Bobby learns bitwise operations

```python
import numpy as np
```

```python
keywords = ["AND", "OR", "LSHIFT", "RSHIFT", "NOT"]
```

```python
circuit = [
    "123 -> x",
    "456 -> y",
    "x AND y -> d",
    "x OR y -> e",
    "x LSHIFT 2 -> f",
    "y RSHIFT 2 -> g",
    "NOT x -> h",
    "NOT y -> i",
]
```

```python
def contains_keyword(line):
    for k in keywords:
        if k in line:
            return True
    return False

def initialize_circuit(circuit):
    signals = {}
    for line in circuit:
        # print(line)
        if contains_keyword(line):
            # print("found keyword")
            continue
        else:
            # print("adding...")
            a, _, c = line.split()
            signals[c] = np.uint16(a)
    return signals

def apply_instructions(signals, circuit):
    for line in circuit:
        if "AND" in line:
            a, _, b, _, c = line.split()
            signals[c] = signals[a] & signals[b]
        elif "OR" in line:
            a, _, b, _, c = line.split()
            signals[c] = signals[a] | signals[b]
        elif "LSHIFT" in line:
            a, _, by, _, c = line.split()
            signals[c] = signals[a] << np.uint16(by)
        elif "RSHIFT" in line:
            a, _, by, _, c = line.split()
            signals[c] = signals[a] >> np.uint16(by)
        elif "NOT" in line:
            _, a, _, c = line.split()
            signals[c] = ~signals[a]

    return signals
```

```python
signals = initialize_circuit(circuit)
signals = apply_instructions(signals, circuit)
signals
```
