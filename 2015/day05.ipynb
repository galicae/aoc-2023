# parsing the naughty list

Santa has complicated rules about who's naughty and no help in figuring it out.

> A nice string is one with all of the following properties:
> 
> - It contains at least three vowels (aeiou only), like aei, xazegov, or aeiouaeiouaeiou.
> - It contains at least one letter that appears twice in a row, like xx, abcdde (dd), or aabbccdd (aa, bb, cc, or dd).
> - It does not contain the strings ab, cd, pq, or xy, even if they are part of one of the other requirements.


```python
from util import file_to_list
```

```python
VOWELS = "aeiou"
NAUGHTY = ["ab", "cd", "pq", "xy"]
```

```python
def double_letter(word):
    for i, letter in enumerate(word[:-1]):
        if letter == word[i + 1]:
            return True
    return False

def enough_vowels(word, enough=3):
    vowel_counter = 0
    for i, letter in enumerate(word):
        if letter in VOWELS:
            vowel_counter += 1
        if vowel_counter >= enough:
            return True
    return False

def is_naughty(word, bad_words=NAUGHTY):
    for w in bad_words:
        if w in word:
            return True
    return False

```

```python
word = "ugknbfddgicrmopn"
assert not is_naughty(word)
assert double_letter(word)
assert enough_vowels(word)
```

```python
word = "aaa"
assert not is_naughty(word)
assert double_letter(word)
assert enough_vowels(word)
```

```python
word = "jchzalrnumimnmhp"
assert not is_naughty(word)
assert not double_letter(word)
```

```python
def nice(word):
    return enough_vowels(word) and double_letter(word) and not is_naughty(word)
```

```python
names_list = file_to_list('./data/day05.txt')
```

```python
total_nice = 0
for kid in names_list:
    if nice(kid):
        total_nice += 1

print(total_nice)
```

## new nice rules

> Now, a nice string is one with all of the following properties:
> 
> - It contains a pair of any two letters that appears at least twice in the string without overlapping, like xyxy (xy) or aabcdefgaa (aa), but not like aaa (aa, but it overlaps).
> - It contains at least one letter which repeats with exactly one letter between them, like xyx, abcdefeghi (efe), or even aaa.


```python
word = "qjhvhtzxzqqjkmpb"
```

> pair of any two letters that appears twice, non-overlapping

--> find and count all possible substrings without overlaps

```python
def all_substrings(word, slen=2):
    substrings = []
    for i in range(len(word)-slen):
        substrings.append(word[i:i+slen])
    return substrings

def repeated_substrings(word, slen=2):
    substrings = all_substrings(word, slen=slen)
    for s in substrings:
        freq = word.count(s)
        if freq > 1:
            return True
    return False
```

```python
assert repeated_substrings(word)
```

> one letter which repeats with exactly one letter between them

--> so if we ignore every second letter we need repeating letters; we can just repurpose the
`double_letter` rule

```python
def double_letter_interleaved(word):
    return double_letter(word[::2]) or double_letter(word[1::2])
```

```python
assert double_letter_interleaved(word)
```

```python
def new_nice(word):
    return repeated_substrings(word) and double_letter_interleaved(word)
```

```python
total_nice = 0
for kid in names_list:
    if new_nice(kid):
        total_nice += 1

print(total_nice)
```
